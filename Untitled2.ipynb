```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### 2. Group Data by Customer and Calculate Summary Metrics

```python
# Group by 'Customer Name' and calculate total sales, total profit, and number of orders
customer_summary = df.groupby('Customer Name').agg(
    total_sales=('Sales', 'sum'),
    total_profit=('Profit', 'sum'),
    number_of_orders=('Order ID', 'nunique') # Use nunique to count unique orders
).reset_index()

# Display the summary for the first few customers
display(customer_summary.head())
```

### 3. Find Top 10 Customers by Total Sales

```python
# Sort the customer_summary DataFrame by 'total_sales' in descending order
top_10_customers_by_sales = customer_summary.sort_values(by='total_sales', ascending=False)

# Select the top 10 customers
top_10_customers_by_sales = top_10_customers_by_sales.head(10)

# Display the top 10 customers by total sales
display(top_10_customers_by_sales)
```

### 4. Find Top 10 Customers by Number of Orders (Frequency)

```python
# Sort the customer_summary DataFrame by 'number_of_orders' in descending order
top_10_customers_by_orders = customer_summary.sort_values(by='number_of_orders', ascending=False)

# Select the top 10 customers
top_10_customers_by_orders = top_10_customers_by_orders.head(10)

# Display the top 10 customers by number of orders
display(top_10_customers_by_orders)
```

### 5. Compare Performance by Customer Segment (Sales and Profit)

```python
# Group by 'Segment' and calculate total sales and total profit
segment_performance = df.groupby('Segment').agg(
    total_sales=('Sales', 'sum'),
    total_profit=('Profit', 'sum')
).reset_index()

# Display the segment performance
display(segment_performance)
```

From the `segment_performance` table above, we can identify:

*   **Segment with most sales:** The segment with the highest value in the `total_sales` column.
*   **Segment with most profit:** The segment with the highest value in the `total_profit` column.

### 6. Create 2 Visualizations

#### Total Sales by Customer Segment

```python
# Create a bar plot for Total Sales by Segment
plt.figure(figsize=(10, 6))
sns.barplot(x='Segment', y='total_sales', data=segment_performance, palette='viridis')
plt.title('Total Sales by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Sales ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

#### Total Profit by Customer Segment

```python
# Create a bar plot for Total Profit by Segment
plt.figure(figsize=(10, 6))
sns.barplot(x='Segment', y='total_profit', data=segment_performance, palette='magma')
plt.title('Total Profit by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Profit ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
